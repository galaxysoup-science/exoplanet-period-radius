```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data (upload the CSV to Colab first)
df = pd.read_csv("/PSCompPars_2026.11.12_15.36.26.csv", comment="#")

# Keep only rows that have both values
df = df.dropna(subset=["pl_orbper", "pl_rade"])

# Make the plot
plt.figure(figsize=(8, 6))
plt.scatter(df["pl_orbper"], df["pl_rade"], s=5, alpha=0.5)
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Orbital period (days)")
plt.ylabel("Planet radius (Earth radii)")
plt.title("Known exoplanets: period vs radius")
plt.savefig("period_vs_radius.png", dpi=200)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
for method, group in df.groupby("discoverymethod"):
    plt.scatter(group["pl_orbper"], group["pl_rade"], s=5, alpha=0.6, label=method)
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Orbital period (days)")
plt.ylabel("Planet radius (Earth radii)")
plt.title("Exoplanets by discovery method")
plt.legend(markerscale=3, fontsize=8)
plt.savefig("by_method.png", dpi=200)
plt.show()
```
